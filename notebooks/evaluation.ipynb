```python

import sys


appand_path = 'Agentic-Medical-Insights'
if appand_path not in sys.path:
   sys.path.append(appand_path)
from app.core.pipeline import MedicalInsightsPipeline
import asyncio

SAMPLE_REPORT = 'sample.pdf'

with open(SAMPLE_REPORT, 'rb') as f:
    file_bytes = f.read()

pipeline = MedicalInsightsPipeline()

initial_state = {'file_bytes': file_bytes, 'filename': 'sample.pdf'}
state = await pipeline.ocr_node(initial_state)

print('Pipeline completed!')
print('Stage:', state.get('current_stage' ))
print('Report type:', state.get('report_type'))
print('OCR Markdown:')
print(state.get('ocr_markdown'))
print('callable:', callable(state.get('ocr_markdown')))
print(state.get('ocr_markdown'))



```
